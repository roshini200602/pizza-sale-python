## Data cleaning and preproscessing

```python
#check for null values
print(df.isnull().sum())
```

```python
#check for duplicates
print(df.duplicated().sum())
```

```python
df.dtypes
```

```python
#cover order_date and order_time  ---- making them proper --covert it to date time format
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True,format='mixed')
```

```python
df['order_time'] = df['order_time'].str.replace('.', ':')
df['order_time']=pd.to_datetime(df['order_time'],format='%H:%M:%S')
```

```python
#extracting month,date and year from order_date

#year
df['year']=df['order_date'].dt.year
```

```python
#month-- month number-- dt.month
#month name---dt.month_namne()

df['month']=df['order_date'].dt.month
df['month_name']=df['order_date'].dt.month_name()
```

```python
df
```

```python
#day
#day number-- dt.day
#day name --- dt.day_name(?)

df["day_number"]=df['order_date'].dt.day
df["day_name"]=df['order_date'].dt.day_name()
```

```python
df
```

```python
#exracting hour,minutes,seconds

df['hour']=df['order_time'].dt.hour
```

```python
#df['minute']=df['order_time'].dt.minute
#df['second']=df['orde_time'].dt.second
```

```python
df['quarter']=df['order_date'].dt.quarter
```

```python
df
```

```python
#remove unncecssary column -- year,month_number,day_number
df.drop(["year","month","day_number"],axis=1,inplace=True)
```

```python
#rename columns
print(df.rename(columns={'pizza_ingredients':'Ingredients','pizza_name_id':'Name_id'}))  #inplace=True(inside rename function)
```

## Basic visulizations

```python
#order in each pizza size
a=df['pizza_size'].value_counts()
print(a)
x=a.index
y=a.values
print(x)
print(y)
plt.bar(x,y)
plt.show()
```

```python
#order in each pizza name
pizza_orders = df['pizza_name'].value_counts()
pizza_orders.plot(
    kind='pie',
    figsize=(10,10),
    autopct='%1.1f%%'
)
plt.title('Number of Orders for Each Pizza Name')
plt.ylabel('')
plt.show()

```

```python
#order in each pizza category
x = df['pizza_category'].value_counts()

m=x.index
n=x.values
plt.pie(n,labels=m,autopct="%1.2f%%")
plt.title('Number of Orders in Each Pizza Category')
plt.show()
```

```python
#distribution of total price-- histogram
a=df["total_price"]
k=[0,5,10,15,20,25,30,35,45,50]
plt.hist(a,bins=k)
plt.title("Distribution of total price")
plt.xlabel("Total price")
plt.ylabel("Count")
#plt.grid()
plt.show()
```

```python
#Distribution of unit price
l=df['unit_price'].head(100) # head is used to get first 100 order or data in list
plt.plot(l,"b.:")
plt.ylabel("unit price")
plt.xlabel("count")
plt.title("Distribution of unit price")
plt.show()
```

## Groupby analysis

```python
#Total pizza saled in every category

a=df.groupby('pizza_category')['quantity'].sum()
print(a)
```

```python
#Total quantity sold and total revenue in each pizza category
a=df.groupby('pizza_category')[['quantity','total_price']].sum()
print(a)
```

```python
#Total orders in each quantity (based on order id)
k=df.groupby("quantity")['order_id'].nunique()
print(k)
```

```python
k=df.groupby("quantity")['order_id'].count()
print(k)
```

```python
#calcute total revenue for each month
m=df.groupby("month_name")['total_price'].sum()
print(m)
```

```python
#calculate total orders for every day of week
m=df.groupby("day_name")['order_id'].nunique()
print(m)
```

```python
#calculate total quantity sold in every hours
k=df.groupby("hour")['quantity'].sum()
print(k)
m=k.index
n=k.values
plt.plot(m,n,"gd--")
plt.title("total quantity sold in every hour")
plt.xlabel("hour")
plt.ylabel("total quqntity sold")
plt.show()
```

```python
#top 5 pizza based on revenue
top=df.groupby("pizza_name")['total_price'].sum().sort_values(ascending=False).head(5)
print(top)
```

```python
x=top.index
y=top.values
plt.barh(x,y,color="green")
plt.title("Top 5 pizza based on revenue") 
plt.xlabel("total revenue")
plt.ylabel("pizza name")
plt.show()
```

```python
#find orders based on quantity
n=df.groupby('quantity')['order_id'].nunique()
print(n)
```

```python
x=n.index
y=n.values
plt.pie(y,labels=x,autopct="%1.2f%%")
plt.title('Number of Orders in Each quantity')
plt.show()
```

```python
#revenue generated by each pizza
r=df.groupby('pizza_name')['total_price'].sum()
print(r)
```

```python
#top 10 sale pizza by sale quantity
top10=df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).head(10)
print(top10)
```

```python
#bottom 10 pizza by sale quantity
bottom10=df.groupby('pizza_name')['quantity'].sum().sort_values().tail(10)
print(bottom10)
```

```python
#top 5 pizza by revenue
t5=df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(10)
print(t5)
```

```python
#bottom 5 pizza by revenue
b5=df.groupby('pizza_name')['total_price'].sum().sort_values().tail(5)
print(b5)
```

```python
#find highest revenue generating category
x=df.groupby('pizza_category')['total_price'].sum().sort_values(ascending=False).head(1)
print(x)
```

```python
#find highest revenue generation size
s=df.groupby('pizza_size')['total_price'].sum().sort_values(ascending=False).head(1)
print(s)
```

```python
#find highest ordered size
o=df.groupby("pizza_size")['order_id'].nunique().sort_values(ascending=False).head(1)
print(o)
```

```python
#total revenue by day of week
day=df.groupby('day_name')['total_price'].sum()
print(day)
```

```python
x=day.index
y=day.values
plt.plot(x,y,'go--')
plt.title("total revenue by day of week")
plt.xlabel("week name")
plt.ylabel("revenue")
plt.show()
```

```python
#find busiest day
b=df.groupby('day_name')['order_id'].nunique().sort_values(ascending=False).head(1)
print(b)
```

```python
#KPI's
#Total orders

t_order=df['order_id'].nunique()
print("Total order:",t_order)
```

```python
#total revenue
t_revenue=df['total_price'].sum()
print("total revenue:",t_revenue)
```

```python
#total quantity sold
t_sold=df['quantity'].sum()
print("total quantity sold:",t_sold)
```

```python
#average price per order
#total_revenue/total_order
avg_price=df['total_price'].sum()/df['order_id'].nunique()
print("average price per order:",avg_price)
print("average price per order:",round(avg_price,2))
```

```python
#average pizza per order
avg_order=df['quantity'].sum()/df['order_id'].nunique()
print("average pizza per order:",avg_order)
print("average pizza per order:",round(avg_order,2))
```
